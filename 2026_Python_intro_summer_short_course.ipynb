```python
# functions are just lists of commands
# they can take multiple 'arguments' (in parentheses) as inputs
# they can 'return' an answer, which might be a number, string, list, ...

def text_repeater(text, reps):
  for i in range(reps):
    print(text)

text_repeater("oh it's you", 4)

text_repeater("Yes, it's me", 2)

def another_function():
  print('')
  print('this')
  print('that')
  print(2*3)

another_function()

def square_it(x):
  return x**2

print('')
z = 5
print( z, square_it(z))
```

```python
import time
# Using a loop
start_time_loop = time.time()
my_list = list(range(10000000))
squared_list = []
for x in my_list:
    squared_list.append(x**2)
#squared_list = [x**2 for x in my_list]
end_time_loop = time.time()
print(f"Loop time: {end_time_loop - start_time_loop:.6f} seconds")

for x in squared_list[0:10]:
  print(x)
```

```python
# prompt: import numpy and do a speed comparison between a numpy array operation and a loop raising each member of a list to the second power

import numpy as np

# Using NumPy
start_time_numpy = time.time()
my_array = np.arange(10000000)  # note difference from array
squared_array = my_array**2
end_time_numpy = time.time()
print(f"NumPy time: {end_time_numpy - start_time_numpy:.6f} seconds")

for x in squared_array[0:10]:
  print(x)
```

**Comparisons by "If" statements**

```python
# Is your drink 'small', 'medium', or 'large'
size = 16 # ounces - but note that the computer does not know about physical units unless you tell it

def sizer(size):
  # these comparisons are performed in order and they stop once the condition is true
  if size < 12:
    print(size, '= small')
  elif size < 16:  # elif abbreivates "else if"
    print(size, '= medium')
  else:
    print(size, '= large')

# note that '<' is different from '<=' (less than or equal)

sizer(size)
```

**Introduction to pandas**

```python
import pandas as pd # pandas abbreviates 'panel data analysis', economics jargon

# Creating a DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Ernesto', 'Frieda'],
    'Age': [24, 27, 22, 32, 81, 16],
    'Score': [88, 92, 95, 89, 23, 98]
}
df = pd.DataFrame(data)

print('DataFrame:')
print(df)

df.head() # useful for big datasets - shows first 5 rows
```

```python
df.tail() # shows last 5 rows
```

```python
# Basic operations on DataFrame
mean_age = df['Age'].mean()  # function 'mean' is built into pandas
max_score = df['Score'].max() # so is function 'max'

print('Mean Age:', mean_age)
print('Max Score:', max_score)
```

```python
# Adding a new column
passing_criteria = 90
df['Pass'] = df['Score'] > passing_criteria  # analogous to if/elif statement
print('Updated DataFrame:')
print(df)
```

```python
# prompt: create a list with the age column of df

age_list = df['Age'].tolist()
print(age_list)

# create a sorted version of the list without changing the original
sorted_age_list = sorted(age_list)
print(sorted_age_list)
print(age_list)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Generate hypothetical data for Spain's games in the 2026 FIFA World Cup
# This is synthetic data as the 2026 World Cup has not occurred.
data = {
    'Date': pd.to_datetime(['2026-06-15', '2026-06-20', '2026-06-25', '2026-06-30', '2026-07-05', '2026-07-09']),
    'Opponent': ['Canada', 'Morocco', 'Japan', 'Germany', 'Brazil', 'France'],
    'Spain_Goals': [2, 1, 3, 2, 1, 0],
    'Opponent_Goals': [0, 1, 1, 2, 0, 1]
}
df_spain_games = pd.DataFrame(data)

# Sort by date to ensure the time series is correct
df_spain_games = df_spain_games.sort_values(by='Date')

# Calculate cumulative goals for Spain
df_spain_games['Cumulative_Goals'] = df_spain_games['Spain_Goals'].cumsum()

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(df_spain_games['Date'], df_spain_games['Cumulative_Goals'], marker='o', linestyle='-')

# Add labels and title
plt.title('Total Goals Scored by Spain Over Time (2026 FIFA World Cup - Hypothetical Data)')
plt.xlabel('Date')
plt.ylabel('Total Goals Scored')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Spain's Total Goals in 2022 FIFA World Cup (Simulated Real Data)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Simulated real data for Spain's games in the 2022 FIFA World Cup
# This data is based on Spain's actual 2022 World Cup matches for demonstration.
data_2022 = {
    'Match_Date': pd.to_datetime(['2022-11-23', '2022-11-27', '2022-12-01', '2022-12-06']),
    'Opponent': ['Costa Rica', 'Germany', 'Japan', 'Morocco'],
    'Spain_Goals': [7, 1, 1, 0],
    'Opponent_Goals': [0, 1, 2, 0]
}
df_spain_2022 = pd.DataFrame(data_2022)

# Sort by date to ensure the time series is correct
df_spain_2022 = df_spain_2022.sort_values(by='Match_Date')

# Calculate cumulative goals for Spain
df_spain_2022['Cumulative_Goals'] = df_spain_2022['Spain_Goals'].cumsum()

# Display the simulated data
print("Simulated 2022 World Cup Data for Spain:")
display(df_spain_2022)
```

```python
# Create the plot for 2022 data
plt.figure(figsize=(10, 6))
plt.plot(df_spain_2022['Match_Date'], df_spain_2022['Cumulative_Goals'], marker='o', linestyle='-', color='red')

# Add labels and title
plt.title('Total Goals Scored by Spain Over Time (2022 FIFA World Cup - Simulated Data)')
plt.xlabel('Match Date')
plt.ylabel('Cumulative Goals Scored')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
